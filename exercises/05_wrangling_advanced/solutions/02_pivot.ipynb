# 正解 — Must 02 — 縦横変換

ペンギンデータをmeltで縦持ちにし、pivot_tableで横に戻してshapeを比較してください。

## 準備


```python
import pandas as pd

df = pd.read_parquet("../../../data/penguins.parquet")
print(f"{df.shape=}")

df_wide = df.loc[
    :,
    ["Individual_ID", "Species_short", "Culmen_Length", "Flipper_Length", "Body_Mass"],
].dropna()
print(f"{df_wide.shape=}")
```

## TODO

`df_wide`を`Individual_ID`と`Species_short`をキーに縦持ちのデータフレーム`df_long`に変換してください。カラム名を格納するカラムを`measure`、値を格納するカラムを`value`とします。

```python
df_long = df_wide.melt(
    id_vars=["Individual_ID", "Species_short"],
    var_name="measure",
    value_name="value",
)
print(f"{df_long.shape=}")
print(f"{df_long.head()=}")
```

`df_long`を`Individual_ID`と`Species_short`をキーに横持ちのデータフレーム`df_wide_back`に変換してください。カラム名は`measure`カラムから、値は`value`カラムから取ります。

```python
df_wide_back = df_long.pivot_table(
    index=["Individual_ID", "Species_short"],
    columns="measure",
    values="value",
    observed=True,
).reset_index()  # これがないと、indexがMultiIndexになる
print(f"{df_wide_back.shape=}")
print(f"{df_wide_back.head()=}")
```

`df_wide`と`df_wide_back`は行数が異なります。その原因を調べてください。

```python
print(f"{df_wide.shape=}, {df_wide_back.shape=}")
print(f"{df_wide.loc[:, ['Individual_ID', 'Species_short']].describe()=}")
print(f"{df_wide_back.loc[:, ['Individual_ID', 'Species_short']].describe()=}")
print(f"{df_wide.loc[:, ['Individual_ID', 'Species_short']].value_counts()=}")
print(f"{df_wide_back.loc[:, ['Individual_ID', 'Species_short']].value_counts()=}")
# 原因は、縦横変換のキー（indexに指定したIndividual_IDとSpecies_shortの組）に重複があること。
# キーに重複があるとpd.pivot_table()関数の引数aggfuncに指定した関数（デフォルトは"mean"）で集約される。

# 参考：pd.pivot()関数はキーがユニークである必要があり、重複があるとエラーになる。
# df_long.pivot(
#     index=["Individual_ID", "Species_short"],
#     columns="measure",
#     values="value",
# )
```
