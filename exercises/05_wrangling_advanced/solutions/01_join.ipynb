# 正解 — Must 01 — 結合（merge）

`Island`をキーに`df_chinstrap_gentoo`と`df_biscoe_torgersen`を結合し、left / inner / right / outer の結果を確認してください。

`python/` に入って実行してください。


## 準備


```python
import pandas as pd

df = pd.read_parquet("../../../data/penguins.parquet")

# ChinstrapとGentooのデータ
df_chinstrap_gentoo = df.loc[
    df.loc[:, "Species_short"].isin(["Chinstrap", "Gentoo"]),
    ["Species_short", "Island", "Body_Mass"]
]

# BiscoeとTorgersenの個体数
df_biscoe_torgersen = df.loc[
    df.loc[:, "Island"].isin(["Biscoe", "Torgersen"]),
    :
].groupby("Island", as_index=False, observed=True)["Individual_ID"].count()
```

## TODO

`Island`をキーに、`df_chinstrap_gentoo`に`df_biscoe_torgersen`を結合


左結合


```python
df_chinstrap_gentoo.merge(df_biscoe_torgersen, on="Island", how="left")
```

内部結合（共通部分）


```python
df_chinstrap_gentoo.merge(df_biscoe_torgersen, on="Island", how="inner")
```

右結合


```python
df_chinstrap_gentoo.merge(df_biscoe_torgersen, on="Island", how="right")
```

外部結合


```python
df_chinstrap_gentoo.merge(df_biscoe_torgersen, on="Island", how="outer")
```
