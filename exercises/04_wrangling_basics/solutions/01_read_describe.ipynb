penguins.parquetを読み、基本的な確認をしたうえで、列・行の抽出、列の追加、ソートを行ってください。

`python/` に入って実行してください（データは `../../../data/`）。


pandasをimportしてください。


```python
import pandas as pd
```

データをデータフレーム`df`に読み込んでください。


```python
df = pd.read_parquet("../../../data/penguins.parquet")
```

shapeを表示してください。


```python
df.shape
```

各カラムのデータ型を表示してください。


```python
df.dtypes
```

最初の10行を表示してください。


```python
df.head(10)
```

`Culmen_Length`と`Culmen_Depth`の2列を`.loc`で抽出して表示してください。


```python
df.loc[:, ["Culmen_Length", "Culmen_Depth"]]
```

`Date_Egg`列を`.loc`で抽出し、Seriesとして表示してください。


```python
df.loc[:, "Date_Egg"]
```

`assign`で`flipper_bill_ratio`（`Flipper_Length / Culmen_Length`）を追加したDataFrameの先頭を表示してください。


```python
df.assign(
  flipper_bill_ratio=lambda d:
    d.loc[:, "Flipper_Length"] /
    d.loc[:, "Culmen_Length"]
).head()
```

`Sex`が`MALE`かつ`Island`が`Biscoe`の行を`.loc`で抽出し、件数と先頭を表示してください。


```python
df_male_biscoe = df.loc[
  (df.loc[:, "Sex"] == "MALE") &
  (df.loc[:, "Island"] == "Biscoe")
]
print(df_male_biscoe.shape)
df_male_biscoe.head()
```

`Body_Mass`の昇順でソートし、先頭を表示してください。


```python
df.sort_values("Body_Mass").head()
```

`Body_Mass`昇順・`Culmen_Length`降順でソートし、先頭を表示してください。


```python
df.sort_values(
  ["Body_Mass", "Culmen_Length"],
  ascending=[True, False]
).head()
```
