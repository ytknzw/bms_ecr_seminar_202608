# 正解 — Must 02 — groupby

島×種の平均体重、および種ごとの件数・平均体重・ひれ長の中央値を集計してください。

スライドのgroupby.aggを発展させる。`python/` に入って実行してください。


## 準備


```python
import pandas as pd

df = pd.read_parquet("../../../data/penguins.parquet")
```

## TODO

島と種の組ごとの`Body_Mass`の平均を計算してください。

```python
df.groupby(["Island", "Species_short"], observed=True)["Body_Mass"].mean()
```

種ごとの`Body_Mass`のサンプルサイズと平均、`Flipper_Length`の中央値を計算して、平均体重の大きい順に表示してください。

```python
df.groupby("Species_short", observed=True).agg(
    n=("Body_Mass", "size"),
    body_mass_mean=("Body_Mass", "mean"),
    flipper_mean=("Flipper_Length", "median"),
).sort_values("body_mass_mean", ascending=False)
```
