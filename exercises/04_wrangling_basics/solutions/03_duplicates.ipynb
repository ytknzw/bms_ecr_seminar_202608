# 正解 — Must 03 — 重複の確認

Adelieについて、件数・ユニーク数・完全重複・Individual_IDの重複を調べてください。

`python/` に入って実行してください（データは `../../../data/`）。


## 準備

```python
import pandas as pd

df = pd.read_parquet("../../../data/penguins.parquet")
```

`Species_short`が`Adelie`の行だけを抽出し、`df_adelie`に入れてください。

```python
df_adelie = df.loc[df["Species_short"] == "Adelie", :]
```

`df_adelie`のshapeを表示してください。

```python
df_adelie.shape
```

`df_adelie`の各列のユニーク数を`nunique(dropna=False)`で数えてください。

```python
df_adelie.nunique(dropna=False)
```

完全重複行の件数を`duplicated()`で数えてください。

```python
df_adelie.duplicated().sum()
```

`Individual_ID`が重複している行の件数を計算してください。

```python
is_dup_id = df_adelie.duplicated(subset=["Individual_ID"], keep=False)
is_dup_id.sum()
```

その重複行を`Individual_ID`でソートし、先頭を表示してください。

```python
df_adelie.loc[is_dup_id, :].sort_values("Individual_ID").head()
```
