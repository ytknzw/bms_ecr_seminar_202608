`Culmen_Length` / `Culmen_Depth` / `Flipper_Length` / `Body_Mass`から`Species_short`を`KNeighborsClassifier()`で予測し、テストデータのaccuracyとclassification_reportを表示してください。

`python/` に入って実行してください（データは `../../../data/`）。


## 準備

```python
import pandas as pd
from sklearn.metrics import accuracy_score, classification_report
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

df = pd.read_parquet("../../../data/penguins.parquet").dropna(
    subset=["Culmen_Length", "Culmen_Depth", "Flipper_Length", "Body_Mass", "Species_short"]  # 欠損を除去
)
```

特徴量`X`（`Culmen_Length`, `Culmen_Depth`, `Flipper_Length`, `Body_Mass`）と目的変数`y`（`Species_short`）を用意してください。

```python
X = df.loc[:, ["Culmen_Length", "Culmen_Depth", "Flipper_Length", "Body_Mass"]]
y = df.loc[:, "Species_short"]
```

`train_test_split()`関数で学習データとテストデータに分けてください。比率は7:3、乱数シードは123とします。

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=123
)
```

`KNeighborsClassifier()`でモデルを構築してください。

```python
model_cls = KNeighborsClassifier()
model_cls.fit(X_train, y_train)
```

テストデータで予測してください。

```python
y_predicted = model_cls.predict(X_test)
```

テストデータについてaccuracyとclassification_reportを表示してください。

```python
print(f"{accuracy_score(y_test, y_predicted)=}")
print(f"{classification_report(y_test, y_predicted)=}")
```
