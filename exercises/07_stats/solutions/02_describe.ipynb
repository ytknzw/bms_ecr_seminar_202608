# 正解 — Must 02 — 要約・分位点・欠測

penguins.parquetを読み、定量列の要約・Body_Massの分位点・欠測の多い列を確認してください。

データは `../../../data/`にあります。

## 準備

```python
import pandas as pd

df = pd.read_parquet("../../../data/penguins.parquet")
```

`Culmen_Length`, `Culmen_Depth`, `Flipper_Length`, `Body_Mass`の各カラムの要約統計量を`describe()`メソッドで表示してください。


```python
df.loc[:, ["Culmen_Length", "Culmen_Depth", "Flipper_Length", "Body_Mass"]].describe()
```

`Body_Mass`カラムの5,50,95,99パーセンタイル点の値を表示してください。


```python
df.loc[:, "Body_Mass"].quantile([0.05, 0.5, 0.95, 0.99])
```

欠測値の多いカラム上位5つを表示してください。


```python
df.isna().mean().sort_values(ascending=False).head(5)
```
