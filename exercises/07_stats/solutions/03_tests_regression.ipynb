# 正解 — Must 03 — t検定と回帰

Sex間のBody_MassのWelchのt検定と、Body_Mass ~ Culmen_Length + Flipper_LengthのOLSを行い結果を表示してください。

## 準備

```python
import pandas as pd
import statsmodels.formula.api as smf
from scipy import stats
```

penguins.parquetを読み込み、`Body_Mass`, `Sex`, `Culmen_Length`, `Flipper_Length`のいずれかに欠損のある行を除いてください。

ヒント：欠損のある行の除去→`dropna()`メソッド

```python
df = pd.read_parquet("../../../data/penguins.parquet")
df = df.dropna(
    how="any",
    subset=["Body_Mass", "Sex", "Culmen_Length", "Flipper_Length"],
)
```

群を`Sex`で分けて、`Body_Mass`についてWelchのt検定を行い、t値、p値、各群の平均値を表示してください。

```python
female = df.loc[df.loc[:, "Sex"].astype(str) == "FEMALE", "Body_Mass"]
male = df.loc[df.loc[:, "Sex"].astype(str) == "MALE", "Body_Mass"]
t, p = stats.ttest_ind(female, male, equal_var=False)
print(f"{t=:.3f}, {p=:.4g}")
print(f"{female.mean()=:.1f}, {male.mean()=:.1f}")
```

モデル`Body_Mass ~ Culmen_Length + Flipper_Length`をOLSで推定し、結果の`summary().tables[1]`を表示してください。

```python
model = smf.ols(
    "Body_Mass ~ Culmen_Length + Flipper_Length",
    data=df,
).fit()
print(f"{model.summary().tables[1]=}")
```
