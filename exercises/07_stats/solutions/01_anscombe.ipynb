必要なライブラリーをimportしてください。

```python
import pandas as pd
import plotly.express as px
```

`../../../data/anscombe.csv`を読み込んで`df_org`に入れてください。

```python
df_org = pd.read_csv("../../../data/anscombe.csv")
```

`df_org`の`shape`を表示してください。

```python
df_org.shape
```

`df_org`の先頭10行を表示してください。

```python
df_org.head(10)
```

`value_counts()`メソッドで`rownames`の各要素の出現数を数えてください。

```python
df_org.loc[:, "rownames"].value_counts()
```

以下の手順で`df_org`（`x1`,`y1`,...）を縦持ち（`dataset`, `x`, `y`の3列）の`df`に整形してください。
1. `df_org`の`x1`〜`y4`を縦持ちに
1. `x1`〜`y4`を入れた列を文字（`x`, `y`）と数字（`1`〜`4`）に分割（数字が`dataset`を表す）
    - `str.extract()`メソッドに正規表現`r"^([xy])([1234])$"`を渡す。
1. `x`と`y`を横持ちに
1. `rownames`を落とす

```python
# 1
df_long = df_org.melt(id_vars="rownames")

# 2
df_long[["axis", "dataset"]] = df_long["variable"].str.extract(r"^([xy])([1234])$")

# 3, 4
df = df_long.pivot_table(  # キーがユニークになることは明らか
    index=["dataset", "rownames"], columns="axis", values="value"
).reset_index().drop(columns=["rownames"])

print(f"{df.shape=}")
df.head()
```

`dataset`ごとにxとyの平均と分散を計算してください。

```python
df.groupby("dataset", observed=True)[["x", "y"]].agg(["mean", "var"])
```

`dataset`ごとにPearsonの相関係数を計算してください。

ヒント：`groupby()`メソッドは集約キーでグループ化したイテラブルを返す。

```python
for dataset, g in df.groupby("dataset", observed=True):
    corr = g.loc[:, "x"].corr(g.loc[:, "y"])
    print(f"{dataset=}, {corr=}")
```

Plotlyで`dataset`ごとにファセットを分けた散布図を描いてください。

```python
fig = px.scatter(
    df,
    x="x",
    y="y",
    color="dataset",
    facet_col="dataset",
    color_discrete_sequence=px.colors.qualitative.Dark2,
    title="Anscombe's quartet",
)
fig.show()
```
