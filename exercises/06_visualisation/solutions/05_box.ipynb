## 準備（データ読み込み）


```python
import pandas as pd
import plotly.express as px

df = pd.read_parquet("../../../data/penguins.parquet")
```

## TODO

種ごとの`Body_Mass`の箱ひげ図を描いてください。種ごとに色を変えてください。

```python
fig = px.box(
    df,
    x="Species_short",
    y="Body_Mass",
    color="Species_short",
    color_discrete_sequence=px.colors.qualitative.Dark2,
    title="種ごとの体重",
)
fig.show()
```
