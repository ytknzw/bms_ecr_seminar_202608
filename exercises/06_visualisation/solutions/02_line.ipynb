# 正解 — Must 02 — 折れ線

年ごとの平均体重の折れ線グラフを描いてください。

## 準備（データ読み込み）


```python
import pandas as pd
import plotly.express as px

df = pd.read_parquet("../../../data/penguins.parquet")
```

`Date_Egg`から年を取り、種ごとに`Body_Mass`の平均値を集計します。

```python
df_avg_weight_year = df.assign(
    year=df.loc[:, "Date_Egg"].dt.year
).groupby(
    ["Species_short", "year"], as_index=False, observed=True
)["Body_Mass"].mean()
```

## TODO

年ごとの平均体重の折れ線グラフを描いてください。各線と点は種ごとに色と形を変えてください（Dark2、`line_dash`、`symbol`）。縦軸は0始まりにしてください（`fig.update_yaxes(rangemode="tozero")`）。

```python
fig = px.line(
    df_avg_weight_year,
    x="year",
    y="Body_Mass",
    color="Species_short",
    color_discrete_sequence=px.colors.qualitative.Dark2,
    line_dash="Species_short",
    symbol="Species_short",
    title="年ごとの平均体重",
)
fig.update_yaxes(rangemode="tozero")
fig.show()
```
