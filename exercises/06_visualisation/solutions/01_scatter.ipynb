# 正解 — Must 01 — 散布図

散布図をPlotlyで描いてください。

**要件:** Dark2配色と`symbol=Species_short`（CUD）を付けてください。

`python/` に入って実行してください。


## 準備（データ読み込み）


```python
import pandas as pd
import plotly.express as px

df = pd.read_parquet("../../../data/penguins.parquet")
```

## TODO

横軸がくちばしの長さ、縦軸がひれの長さの散布図を描いてください。各データ点は種ごとに色と形を変えてください。

```python
fig = px.scatter(
    df,
    x="Culmen_Length",
    y="Flipper_Length",
    color="Species_short",
    color_discrete_sequence=px.colors.qualitative.Dark2,
    symbol="Species_short",
    title="くちばしの長さとひれの長さ",
)
fig.show()
```

## TODO

横軸がくちばしの長さ、縦軸がひれの長さの散布図を、`facet_col="Species_short"`で種ごとに並べて描いてください。


```python
fig = px.scatter(
    df,
    x="Culmen_Length",
    y="Flipper_Length",
    facet_col="Species_short",
    title="くちばしの長さとひれの長さ（種ごと）",
)
fig.show()
```
