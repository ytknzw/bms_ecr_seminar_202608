# 正解 — Must 04 — ヒストグラム

`Body_Mass`のヒストグラムを描いてください。

## 準備（データ読み込み）


```python
import pandas as pd
import plotly.express as px

df = pd.read_parquet("../../../data/penguins.parquet")
```

## TODO

Body_Massのヒストグラムを描いてください。種ごとに色と模様を変えて（Dark2、`pattern_shape`）、やや色を透明にし（`opacity=0.7`）、重ね書きしてください。


```python
fig = px.histogram(
    df,
    x="Body_Mass",
    color="Species_short",
    color_discrete_sequence=px.colors.qualitative.Dark2,
    pattern_shape="Species_short",
    opacity=0.7,
    barmode="overlay",
    title="体重の分布",
)
fig.show()
```

## TODO

`Body_Mass`のヒストグラムを、`facet_col="Species_short"`で種ごとに並べて描いてください。


```python
fig = px.histogram(
    df,
    x="Body_Mass",
    facet_col="Species_short",
    title="体重の分布（種ごと）",
)
fig.show()
```
